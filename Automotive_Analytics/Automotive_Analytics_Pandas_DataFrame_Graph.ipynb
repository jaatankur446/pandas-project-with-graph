# Pandas DataFrame + Graph Project — Automotive Analytics

**Input files:** `vehicles.csv`, `service_records.csv`

This project contains 30 business scenarios. **Every scenario includes a Pandas DataFrame analysis and a graph.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1=pd.read_csv('vehicles.csv')
df2=pd.read_csv('service_records.csv')
df2['service_date']=pd.to_datetime(df2['service_date'])

df1.head()
df2.head()
```

## Q1. Overview

Display shape, columns, data types and samples.

```python
print(df1.shape, df2.shape)
print(df1.dtypes)
print(df2.dtypes)
df1.head()
df2.head()
df2.select_dtypes('number').mean().plot(kind='bar',figsize=(9,4),title='Average Numeric Columns'); plt.tight_layout(); plt.show()
```

## Q2. Category Count

Count records by city.

```python
r=df2.groupby('city').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(9,5),title='Records by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q3. Category Total

Calculate total service_cost by service_type.

```python
r=df2.groupby('service_type')['service_cost'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Total service_cost by service_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q4. Category Average

Calculate average service_cost by service_type.

```python
r=df2.groupby('service_type')['service_cost'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average service_cost by service_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q5. Top 10

Find top 10 records by service_cost.

```python
r=df2.nlargest(10,'service_cost'); print(r)
r.set_index('vehicle_id')['service_cost'].plot(kind='bar',figsize=(11,5),title='Top 10 service_cost'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q6. Bottom 10

Find bottom 10 records by service_cost.

```python
r=df2.nsmallest(10,'service_cost'); print(r)
r.set_index('vehicle_id')['service_cost'].plot(kind='bar',figsize=(11,5),title='Bottom 10 service_cost'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q7. Missing Values

Calculate missing values in every column and visualize them.

```python
r=df2.isna().sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Missing Values'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q8. Duplicate Check

Count duplicate records and visualize unique versus duplicate rows.

```python
r=pd.Series({'Unique':len(df2)-df2.duplicated().sum(),'Duplicates':df2.duplicated().sum()}); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(6,6),title='Unique vs Duplicate'); plt.ylabel(''); plt.show()
```

## Q9. Distribution

Show descriptive statistics and distribution of service_cost.

```python
print(df2['service_cost'].describe())
df2['service_cost'].plot(kind='hist',bins=20,figsize=(9,5),title='Distribution of service_cost'); plt.xlabel('service_cost'); plt.tight_layout(); plt.show()
```

## Q10. Outliers

Identify service_cost outliers using IQR and visualize them.

```python
q1=df2['service_cost'].quantile(.25); q3=df2['service_cost'].quantile(.75); iqr=q3-q1
out=df2[(df2['service_cost']<q1-1.5*iqr)|(df2['service_cost']>q3+1.5*iqr)]; print('Outliers:',len(out))
df2.boxplot(column='service_cost',figsize=(7,5)); plt.title('service_cost Outliers'); plt.show()
```

## Q11. Monthly Trend

Calculate monthly total service_cost.

```python
df2['service_date']=pd.to_datetime(df2['service_date']); r=df2.groupby(df2['service_date'].dt.to_period('M'))['service_cost'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(11,5),title='Monthly service_cost'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q12. Yearly Trend

Calculate yearly total service_cost.

```python
r=df2.groupby(df2['service_date'].dt.year)['service_cost'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(9,5),title='Yearly service_cost'); plt.tight_layout(); plt.show()
```

## Q13. Percentage Share

Calculate percentage contribution of each service_type to total service_cost.

```python
r=df2.groupby('service_type')['service_cost'].sum(); r=r/r.sum()*100; print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='service_type Share'); plt.ylabel(''); plt.show()
```

## Q14. Cumulative Trend

Calculate cumulative service_cost over time.

```python
r=df2.sort_values('service_date').copy(); r['cumulative']=r['service_cost'].cumsum(); print(r[['service_date','service_cost','cumulative']].head(20))
r.plot(x='service_date',y='cumulative',figsize=(11,5),title='Cumulative service_cost'); plt.tight_layout(); plt.show()
```

## Q15. Category Ranking

Rank service_type by total service_cost.

```python
r=df2.groupby('service_type')['service_cost'].sum().sort_values(ascending=False).reset_index(); r['rank']=r['service_cost'].rank(method='dense',ascending=False); print(r)
r.plot(x='service_type',y='service_cost',kind='bar',figsize=(10,5),title='Category Ranking',legend=False); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q16. Value Segments

Create Low, Medium and High segments for service_cost.

```python
m=df2['service_cost'].mean(); df2['segment']=np.select([df2['service_cost']<m*.7,df2['service_cost']<=m*1.3],['Low','Medium'],default='High'); r=df2['segment'].value_counts(); print(r)
r.plot(kind='bar',figsize=(8,5),title='Value Segments'); plt.tight_layout(); plt.show()
```

## Q17. Cross Tab

Create a pivot table of service_cost using city and service_type.

```python
r=df2.pivot_table(index='city',columns='service_type',values='service_cost',aggfunc='sum',fill_value=0); print(r)
r.plot(kind='bar',figsize=(12,6),title='city vs service_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q18. Entity Totals

Calculate total service_cost by vehicle_id and display top 15.

```python
r=df2.groupby('vehicle_id')['service_cost'].sum().nlargest(15); print(r)
r.plot(kind='bar',figsize=(12,5),title='Top 15 by service_cost'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q19. Entity Frequency

Find the most frequent vehicle_id values.

```python
r=df2['vehicle_id'].value_counts().head(15); print(r)
r.plot(kind='bar',figsize=(11,5),title='Most Frequent vehicle_id'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q20. Inner Join

Merge the two DataFrames using vehicle_id and calculate service_cost by city.

```python
j=df1.merge(df2,on='vehicle_id',how='inner'); r=j.groupby('city')['service_cost'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Joined service_cost by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q21. Join Row Comparison

Compare rows before and after the inner join.

```python
j=df1.merge(df2,on='vehicle_id',how='inner'); r=pd.Series(dict(DF1=len(df1),DF2=len(df2),Joined=len(j))); print(r)
r.plot(kind='bar',figsize=(7,5),title='Join Row Comparison'); plt.ylabel('Rows'); plt.show()
```

## Q22. Left Join Coverage

Measure matching versus unmatched vehicle_id records.

```python
j=df1.merge(df2,on='vehicle_id',how='left',indicator=True); r=j['_merge'].value_counts(); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Join Coverage'); plt.ylabel(''); plt.show()
```

## Q23. Average by Main Category

Calculate average service_cost by city after joining.

```python
j=df1.merge(df2,on='vehicle_id',how='inner'); r=j.groupby('city')['service_cost'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average service_cost by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q24. Monthly Category Trend

Show monthly service_cost trend for each service_type.

```python
j=df2.copy(); j['month']=j['service_date'].dt.to_period('M').astype(str); r=j.pivot_table(index='month',columns='service_type',values='service_cost',aggfunc='sum',fill_value=0); print(r.tail(12))
r.plot(figsize=(12,6),marker='o',title='Monthly service_cost by service_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q25. Top 5 Share

Find top 5 service_type categories by total service_cost.

```python
r=df2.groupby('service_type')['service_cost'].sum().nlargest(5); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Top 5 service_type Share'); plt.ylabel(''); plt.show()
```

## Q26. Correlation

Calculate and visualize correlations among numeric columns.

```python
r=df2.select_dtypes('number').corr(); print(r)
plt.figure(figsize=(9,7)); plt.imshow(r,aspect='auto',cmap='viridis'); plt.colorbar(); plt.xticks(range(len(r)),r.columns,rotation=60); plt.yticks(range(len(r)),r.columns); plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

## Q27. KPI Summary

Create total, average, minimum and maximum KPIs for service_cost.

```python
r=pd.Series(dict(Total=df2['service_cost'].sum(),Average=df2['service_cost'].mean(),Minimum=df2['service_cost'].min(),Maximum=df2['service_cost'].max())); print(r)
r.plot(kind='bar',figsize=(8,5),title='service_cost KPI Summary'); plt.xticks(rotation=0); plt.show()
```

## Q28. Highest vs Lowest

Compare the highest and lowest service_type by total service_cost.

```python
r=df2.groupby('service_type')['service_cost'].sum().sort_values(ascending=False); r=r.iloc[[0,-1]]; print(r)
r.plot(kind='bar',figsize=(7,5),title='Highest vs Lowest service_type'); plt.xticks(rotation=30); plt.show()
```

## Q29. Conditional Filter

Show records where service_cost is above its overall average and visualize counts by service_type.

```python
m=df2['service_cost'].mean(); r=df2[df2['service_cost']>m].groupby('service_type').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Above-Average Records by service_type'); plt.xticks(rotation=45); plt.show()
```

## Q30. Final Business Report

Create a final report by service_type with count, total, average, minimum and maximum service_cost.

```python
r=df2.groupby('service_type').agg(records=('service_cost','size'),total=('service_cost','sum'),average=('service_cost','mean'),minimum=('service_cost','min'),maximum=('service_cost','max')).sort_values('total',ascending=False); print(r)
r['total'].plot(kind='bar',figsize=(11,5),title='Final Business Report'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
