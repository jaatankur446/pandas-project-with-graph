# Pandas DataFrame + Graph Project — Real Estate Analytics

**Input files:** `properties.csv`, `property_sales.csv`

This project contains 30 business scenarios. **Every scenario includes a Pandas DataFrame analysis and a graph.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1=pd.read_csv('properties.csv')
df2=pd.read_csv('property_sales.csv')
df2['sale_date']=pd.to_datetime(df2['sale_date'])

df1.head()
df2.head()
```

## Q1. Overview

Display shape, columns, data types and samples.

```python
print(df1.shape, df2.shape)
print(df1.dtypes)
print(df2.dtypes)
df1.head()
df2.head()
df2.select_dtypes('number').mean().plot(kind='bar',figsize=(9,4),title='Average Numeric Columns'); plt.tight_layout(); plt.show()
```

## Q2. Category Count

Count records by city.

```python
r=df2.groupby('city').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(9,5),title='Records by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q3. Category Total

Calculate total sale_price by agent.

```python
r=df2.groupby('agent')['sale_price'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Total sale_price by agent'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q4. Category Average

Calculate average sale_price by agent.

```python
r=df2.groupby('agent')['sale_price'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average sale_price by agent'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q5. Top 10

Find top 10 records by sale_price.

```python
r=df2.nlargest(10,'sale_price'); print(r)
r.set_index('property_id')['sale_price'].plot(kind='bar',figsize=(11,5),title='Top 10 sale_price'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q6. Bottom 10

Find bottom 10 records by sale_price.

```python
r=df2.nsmallest(10,'sale_price'); print(r)
r.set_index('property_id')['sale_price'].plot(kind='bar',figsize=(11,5),title='Bottom 10 sale_price'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q7. Missing Values

Calculate missing values in every column and visualize them.

```python
r=df2.isna().sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Missing Values'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q8. Duplicate Check

Count duplicate records and visualize unique versus duplicate rows.

```python
r=pd.Series({'Unique':len(df2)-df2.duplicated().sum(),'Duplicates':df2.duplicated().sum()}); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(6,6),title='Unique vs Duplicate'); plt.ylabel(''); plt.show()
```

## Q9. Distribution

Show descriptive statistics and distribution of sale_price.

```python
print(df2['sale_price'].describe())
df2['sale_price'].plot(kind='hist',bins=20,figsize=(9,5),title='Distribution of sale_price'); plt.xlabel('sale_price'); plt.tight_layout(); plt.show()
```

## Q10. Outliers

Identify sale_price outliers using IQR and visualize them.

```python
q1=df2['sale_price'].quantile(.25); q3=df2['sale_price'].quantile(.75); iqr=q3-q1
out=df2[(df2['sale_price']<q1-1.5*iqr)|(df2['sale_price']>q3+1.5*iqr)]; print('Outliers:',len(out))
df2.boxplot(column='sale_price',figsize=(7,5)); plt.title('sale_price Outliers'); plt.show()
```

## Q11. Monthly Trend

Calculate monthly total sale_price.

```python
df2['sale_date']=pd.to_datetime(df2['sale_date']); r=df2.groupby(df2['sale_date'].dt.to_period('M'))['sale_price'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(11,5),title='Monthly sale_price'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q12. Yearly Trend

Calculate yearly total sale_price.

```python
r=df2.groupby(df2['sale_date'].dt.year)['sale_price'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(9,5),title='Yearly sale_price'); plt.tight_layout(); plt.show()
```

## Q13. Percentage Share

Calculate percentage contribution of each agent to total sale_price.

```python
r=df2.groupby('agent')['sale_price'].sum(); r=r/r.sum()*100; print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='agent Share'); plt.ylabel(''); plt.show()
```

## Q14. Cumulative Trend

Calculate cumulative sale_price over time.

```python
r=df2.sort_values('sale_date').copy(); r['cumulative']=r['sale_price'].cumsum(); print(r[['sale_date','sale_price','cumulative']].head(20))
r.plot(x='sale_date',y='cumulative',figsize=(11,5),title='Cumulative sale_price'); plt.tight_layout(); plt.show()
```

## Q15. Category Ranking

Rank agent by total sale_price.

```python
r=df2.groupby('agent')['sale_price'].sum().sort_values(ascending=False).reset_index(); r['rank']=r['sale_price'].rank(method='dense',ascending=False); print(r)
r.plot(x='agent',y='sale_price',kind='bar',figsize=(10,5),title='Category Ranking',legend=False); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q16. Value Segments

Create Low, Medium and High segments for sale_price.

```python
m=df2['sale_price'].mean(); df2['segment']=np.select([df2['sale_price']<m*.7,df2['sale_price']<=m*1.3],['Low','Medium'],default='High'); r=df2['segment'].value_counts(); print(r)
r.plot(kind='bar',figsize=(8,5),title='Value Segments'); plt.tight_layout(); plt.show()
```

## Q17. Cross Tab

Create a pivot table of sale_price using city and agent.

```python
r=df2.pivot_table(index='city',columns='agent',values='sale_price',aggfunc='sum',fill_value=0); print(r)
r.plot(kind='bar',figsize=(12,6),title='city vs agent'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q18. Entity Totals

Calculate total sale_price by property_id and display top 15.

```python
r=df2.groupby('property_id')['sale_price'].sum().nlargest(15); print(r)
r.plot(kind='bar',figsize=(12,5),title='Top 15 by sale_price'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q19. Entity Frequency

Find the most frequent property_id values.

```python
r=df2['property_id'].value_counts().head(15); print(r)
r.plot(kind='bar',figsize=(11,5),title='Most Frequent property_id'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q20. Inner Join

Merge the two DataFrames using property_id and calculate sale_price by city.

```python
j=df1.merge(df2,on='property_id',how='inner'); r=j.groupby('city')['sale_price'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Joined sale_price by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q21. Join Row Comparison

Compare rows before and after the inner join.

```python
j=df1.merge(df2,on='property_id',how='inner'); r=pd.Series(dict(DF1=len(df1),DF2=len(df2),Joined=len(j))); print(r)
r.plot(kind='bar',figsize=(7,5),title='Join Row Comparison'); plt.ylabel('Rows'); plt.show()
```

## Q22. Left Join Coverage

Measure matching versus unmatched property_id records.

```python
j=df1.merge(df2,on='property_id',how='left',indicator=True); r=j['_merge'].value_counts(); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Join Coverage'); plt.ylabel(''); plt.show()
```

## Q23. Average by Main Category

Calculate average sale_price by city after joining.

```python
j=df1.merge(df2,on='property_id',how='inner'); r=j.groupby('city')['sale_price'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average sale_price by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q24. Monthly Category Trend

Show monthly sale_price trend for each agent.

```python
j=df2.copy(); j['month']=j['sale_date'].dt.to_period('M').astype(str); r=j.pivot_table(index='month',columns='agent',values='sale_price',aggfunc='sum',fill_value=0); print(r.tail(12))
r.plot(figsize=(12,6),marker='o',title='Monthly sale_price by agent'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q25. Top 5 Share

Find top 5 agent categories by total sale_price.

```python
r=df2.groupby('agent')['sale_price'].sum().nlargest(5); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Top 5 agent Share'); plt.ylabel(''); plt.show()
```

## Q26. Correlation

Calculate and visualize correlations among numeric columns.

```python
r=df2.select_dtypes('number').corr(); print(r)
plt.figure(figsize=(9,7)); plt.imshow(r,aspect='auto',cmap='viridis'); plt.colorbar(); plt.xticks(range(len(r)),r.columns,rotation=60); plt.yticks(range(len(r)),r.columns); plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

## Q27. KPI Summary

Create total, average, minimum and maximum KPIs for sale_price.

```python
r=pd.Series(dict(Total=df2['sale_price'].sum(),Average=df2['sale_price'].mean(),Minimum=df2['sale_price'].min(),Maximum=df2['sale_price'].max())); print(r)
r.plot(kind='bar',figsize=(8,5),title='sale_price KPI Summary'); plt.xticks(rotation=0); plt.show()
```

## Q28. Highest vs Lowest

Compare the highest and lowest agent by total sale_price.

```python
r=df2.groupby('agent')['sale_price'].sum().sort_values(ascending=False); r=r.iloc[[0,-1]]; print(r)
r.plot(kind='bar',figsize=(7,5),title='Highest vs Lowest agent'); plt.xticks(rotation=30); plt.show()
```

## Q29. Conditional Filter

Show records where sale_price is above its overall average and visualize counts by agent.

```python
m=df2['sale_price'].mean(); r=df2[df2['sale_price']>m].groupby('agent').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Above-Average Records by agent'); plt.xticks(rotation=45); plt.show()
```

## Q30. Final Business Report

Create a final report by agent with count, total, average, minimum and maximum sale_price.

```python
r=df2.groupby('agent').agg(records=('sale_price','size'),total=('sale_price','sum'),average=('sale_price','mean'),minimum=('sale_price','min'),maximum=('sale_price','max')).sort_values('total',ascending=False); print(r)
r['total'].plot(kind='bar',figsize=(11,5),title='Final Business Report'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
