# Pandas DataFrame + Graph Project — Cinema Analytics

**Input files:** `movies.csv`, `ticket_sales.csv`

This project contains 30 business scenarios. **Every scenario includes a Pandas DataFrame analysis and a graph.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1=pd.read_csv('movies.csv')
df2=pd.read_csv('ticket_sales.csv')
df2['show_date']=pd.to_datetime(df2['show_date'])

df1.head()
df2.head()
```

## Q1. Overview

Display shape, columns, data types and samples.

```python
print(df1.shape, df2.shape)
print(df1.dtypes)
print(df2.dtypes)
df1.head()
df2.head()
df2.select_dtypes('number').mean().plot(kind='bar',figsize=(9,4),title='Average Numeric Columns'); plt.tight_layout(); plt.show()
```

## Q2. Category Count

Count records by genre.

```python
r=df2.groupby('genre').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(9,5),title='Records by genre'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q3. Category Total

Calculate total box_office by city.

```python
r=df2.groupby('city')['box_office'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Total box_office by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q4. Category Average

Calculate average box_office by city.

```python
r=df2.groupby('city')['box_office'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average box_office by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q5. Top 10

Find top 10 records by box_office.

```python
r=df2.nlargest(10,'box_office'); print(r)
r.set_index('movie_id')['box_office'].plot(kind='bar',figsize=(11,5),title='Top 10 box_office'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q6. Bottom 10

Find bottom 10 records by box_office.

```python
r=df2.nsmallest(10,'box_office'); print(r)
r.set_index('movie_id')['box_office'].plot(kind='bar',figsize=(11,5),title='Bottom 10 box_office'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q7. Missing Values

Calculate missing values in every column and visualize them.

```python
r=df2.isna().sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Missing Values'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q8. Duplicate Check

Count duplicate records and visualize unique versus duplicate rows.

```python
r=pd.Series({'Unique':len(df2)-df2.duplicated().sum(),'Duplicates':df2.duplicated().sum()}); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(6,6),title='Unique vs Duplicate'); plt.ylabel(''); plt.show()
```

## Q9. Distribution

Show descriptive statistics and distribution of box_office.

```python
print(df2['box_office'].describe())
df2['box_office'].plot(kind='hist',bins=20,figsize=(9,5),title='Distribution of box_office'); plt.xlabel('box_office'); plt.tight_layout(); plt.show()
```

## Q10. Outliers

Identify box_office outliers using IQR and visualize them.

```python
q1=df2['box_office'].quantile(.25); q3=df2['box_office'].quantile(.75); iqr=q3-q1
out=df2[(df2['box_office']<q1-1.5*iqr)|(df2['box_office']>q3+1.5*iqr)]; print('Outliers:',len(out))
df2.boxplot(column='box_office',figsize=(7,5)); plt.title('box_office Outliers'); plt.show()
```

## Q11. Monthly Trend

Calculate monthly total box_office.

```python
df2['show_date']=pd.to_datetime(df2['show_date']); r=df2.groupby(df2['show_date'].dt.to_period('M'))['box_office'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(11,5),title='Monthly box_office'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q12. Yearly Trend

Calculate yearly total box_office.

```python
r=df2.groupby(df2['show_date'].dt.year)['box_office'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(9,5),title='Yearly box_office'); plt.tight_layout(); plt.show()
```

## Q13. Percentage Share

Calculate percentage contribution of each city to total box_office.

```python
r=df2.groupby('city')['box_office'].sum(); r=r/r.sum()*100; print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='city Share'); plt.ylabel(''); plt.show()
```

## Q14. Cumulative Trend

Calculate cumulative box_office over time.

```python
r=df2.sort_values('show_date').copy(); r['cumulative']=r['box_office'].cumsum(); print(r[['show_date','box_office','cumulative']].head(20))
r.plot(x='show_date',y='cumulative',figsize=(11,5),title='Cumulative box_office'); plt.tight_layout(); plt.show()
```

## Q15. Category Ranking

Rank city by total box_office.

```python
r=df2.groupby('city')['box_office'].sum().sort_values(ascending=False).reset_index(); r['rank']=r['box_office'].rank(method='dense',ascending=False); print(r)
r.plot(x='city',y='box_office',kind='bar',figsize=(10,5),title='Category Ranking',legend=False); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q16. Value Segments

Create Low, Medium and High segments for box_office.

```python
m=df2['box_office'].mean(); df2['segment']=np.select([df2['box_office']<m*.7,df2['box_office']<=m*1.3],['Low','Medium'],default='High'); r=df2['segment'].value_counts(); print(r)
r.plot(kind='bar',figsize=(8,5),title='Value Segments'); plt.tight_layout(); plt.show()
```

## Q17. Cross Tab

Create a pivot table of box_office using genre and city.

```python
r=df2.pivot_table(index='genre',columns='city',values='box_office',aggfunc='sum',fill_value=0); print(r)
r.plot(kind='bar',figsize=(12,6),title='genre vs city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q18. Entity Totals

Calculate total box_office by movie_id and display top 15.

```python
r=df2.groupby('movie_id')['box_office'].sum().nlargest(15); print(r)
r.plot(kind='bar',figsize=(12,5),title='Top 15 by box_office'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q19. Entity Frequency

Find the most frequent movie_id values.

```python
r=df2['movie_id'].value_counts().head(15); print(r)
r.plot(kind='bar',figsize=(11,5),title='Most Frequent movie_id'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q20. Inner Join

Merge the two DataFrames using movie_id and calculate box_office by genre.

```python
j=df1.merge(df2,on='movie_id',how='inner'); r=j.groupby('genre')['box_office'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Joined box_office by genre'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q21. Join Row Comparison

Compare rows before and after the inner join.

```python
j=df1.merge(df2,on='movie_id',how='inner'); r=pd.Series(dict(DF1=len(df1),DF2=len(df2),Joined=len(j))); print(r)
r.plot(kind='bar',figsize=(7,5),title='Join Row Comparison'); plt.ylabel('Rows'); plt.show()
```

## Q22. Left Join Coverage

Measure matching versus unmatched movie_id records.

```python
j=df1.merge(df2,on='movie_id',how='left',indicator=True); r=j['_merge'].value_counts(); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Join Coverage'); plt.ylabel(''); plt.show()
```

## Q23. Average by Main Category

Calculate average box_office by genre after joining.

```python
j=df1.merge(df2,on='movie_id',how='inner'); r=j.groupby('genre')['box_office'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average box_office by genre'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q24. Monthly Category Trend

Show monthly box_office trend for each city.

```python
j=df2.copy(); j['month']=j['show_date'].dt.to_period('M').astype(str); r=j.pivot_table(index='month',columns='city',values='box_office',aggfunc='sum',fill_value=0); print(r.tail(12))
r.plot(figsize=(12,6),marker='o',title='Monthly box_office by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q25. Top 5 Share

Find top 5 city categories by total box_office.

```python
r=df2.groupby('city')['box_office'].sum().nlargest(5); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Top 5 city Share'); plt.ylabel(''); plt.show()
```

## Q26. Correlation

Calculate and visualize correlations among numeric columns.

```python
r=df2.select_dtypes('number').corr(); print(r)
plt.figure(figsize=(9,7)); plt.imshow(r,aspect='auto',cmap='viridis'); plt.colorbar(); plt.xticks(range(len(r)),r.columns,rotation=60); plt.yticks(range(len(r)),r.columns); plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

## Q27. KPI Summary

Create total, average, minimum and maximum KPIs for box_office.

```python
r=pd.Series(dict(Total=df2['box_office'].sum(),Average=df2['box_office'].mean(),Minimum=df2['box_office'].min(),Maximum=df2['box_office'].max())); print(r)
r.plot(kind='bar',figsize=(8,5),title='box_office KPI Summary'); plt.xticks(rotation=0); plt.show()
```

## Q28. Highest vs Lowest

Compare the highest and lowest city by total box_office.

```python
r=df2.groupby('city')['box_office'].sum().sort_values(ascending=False); r=r.iloc[[0,-1]]; print(r)
r.plot(kind='bar',figsize=(7,5),title='Highest vs Lowest city'); plt.xticks(rotation=30); plt.show()
```

## Q29. Conditional Filter

Show records where box_office is above its overall average and visualize counts by city.

```python
m=df2['box_office'].mean(); r=df2[df2['box_office']>m].groupby('city').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Above-Average Records by city'); plt.xticks(rotation=45); plt.show()
```

## Q30. Final Business Report

Create a final report by city with count, total, average, minimum and maximum box_office.

```python
r=df2.groupby('city').agg(records=('box_office','size'),total=('box_office','sum'),average=('box_office','mean'),minimum=('box_office','min'),maximum=('box_office','max')).sort_values('total',ascending=False); print(r)
r['total'].plot(kind='bar',figsize=(11,5),title='Final Business Report'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
