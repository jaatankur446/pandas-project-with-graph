# Pandas DataFrame + Graph Project — Energy Analytics

**Input files:** `meters.csv`, `energy_usage.csv`

This project contains 30 business scenarios. **Every scenario includes a Pandas DataFrame analysis and a graph.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1=pd.read_csv('meters.csv')
df2=pd.read_csv('energy_usage.csv')
df2['reading_date']=pd.to_datetime(df2['reading_date'])

df1.head()
df2.head()
```

## Q1. Overview

Display shape, columns, data types and samples.

```python
print(df1.shape, df2.shape)
print(df1.dtypes)
print(df2.dtypes)
df1.head()
df2.head()
df2.select_dtypes('number').mean().plot(kind='bar',figsize=(9,4),title='Average Numeric Columns'); plt.tight_layout(); plt.show()
```

## Q2. Category Count

Count records by city.

```python
r=df2.groupby('city').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(9,5),title='Records by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q3. Category Total

Calculate total bill_amount by connection_type.

```python
r=df2.groupby('connection_type')['bill_amount'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Total bill_amount by connection_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q4. Category Average

Calculate average bill_amount by connection_type.

```python
r=df2.groupby('connection_type')['bill_amount'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average bill_amount by connection_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q5. Top 10

Find top 10 records by bill_amount.

```python
r=df2.nlargest(10,'bill_amount'); print(r)
r.set_index('meter_id')['bill_amount'].plot(kind='bar',figsize=(11,5),title='Top 10 bill_amount'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q6. Bottom 10

Find bottom 10 records by bill_amount.

```python
r=df2.nsmallest(10,'bill_amount'); print(r)
r.set_index('meter_id')['bill_amount'].plot(kind='bar',figsize=(11,5),title='Bottom 10 bill_amount'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q7. Missing Values

Calculate missing values in every column and visualize them.

```python
r=df2.isna().sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Missing Values'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q8. Duplicate Check

Count duplicate records and visualize unique versus duplicate rows.

```python
r=pd.Series({'Unique':len(df2)-df2.duplicated().sum(),'Duplicates':df2.duplicated().sum()}); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(6,6),title='Unique vs Duplicate'); plt.ylabel(''); plt.show()
```

## Q9. Distribution

Show descriptive statistics and distribution of bill_amount.

```python
print(df2['bill_amount'].describe())
df2['bill_amount'].plot(kind='hist',bins=20,figsize=(9,5),title='Distribution of bill_amount'); plt.xlabel('bill_amount'); plt.tight_layout(); plt.show()
```

## Q10. Outliers

Identify bill_amount outliers using IQR and visualize them.

```python
q1=df2['bill_amount'].quantile(.25); q3=df2['bill_amount'].quantile(.75); iqr=q3-q1
out=df2[(df2['bill_amount']<q1-1.5*iqr)|(df2['bill_amount']>q3+1.5*iqr)]; print('Outliers:',len(out))
df2.boxplot(column='bill_amount',figsize=(7,5)); plt.title('bill_amount Outliers'); plt.show()
```

## Q11. Monthly Trend

Calculate monthly total bill_amount.

```python
df2['reading_date']=pd.to_datetime(df2['reading_date']); r=df2.groupby(df2['reading_date'].dt.to_period('M'))['bill_amount'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(11,5),title='Monthly bill_amount'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q12. Yearly Trend

Calculate yearly total bill_amount.

```python
r=df2.groupby(df2['reading_date'].dt.year)['bill_amount'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(9,5),title='Yearly bill_amount'); plt.tight_layout(); plt.show()
```

## Q13. Percentage Share

Calculate percentage contribution of each connection_type to total bill_amount.

```python
r=df2.groupby('connection_type')['bill_amount'].sum(); r=r/r.sum()*100; print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='connection_type Share'); plt.ylabel(''); plt.show()
```

## Q14. Cumulative Trend

Calculate cumulative bill_amount over time.

```python
r=df2.sort_values('reading_date').copy(); r['cumulative']=r['bill_amount'].cumsum(); print(r[['reading_date','bill_amount','cumulative']].head(20))
r.plot(x='reading_date',y='cumulative',figsize=(11,5),title='Cumulative bill_amount'); plt.tight_layout(); plt.show()
```

## Q15. Category Ranking

Rank connection_type by total bill_amount.

```python
r=df2.groupby('connection_type')['bill_amount'].sum().sort_values(ascending=False).reset_index(); r['rank']=r['bill_amount'].rank(method='dense',ascending=False); print(r)
r.plot(x='connection_type',y='bill_amount',kind='bar',figsize=(10,5),title='Category Ranking',legend=False); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q16. Value Segments

Create Low, Medium and High segments for bill_amount.

```python
m=df2['bill_amount'].mean(); df2['segment']=np.select([df2['bill_amount']<m*.7,df2['bill_amount']<=m*1.3],['Low','Medium'],default='High'); r=df2['segment'].value_counts(); print(r)
r.plot(kind='bar',figsize=(8,5),title='Value Segments'); plt.tight_layout(); plt.show()
```

## Q17. Cross Tab

Create a pivot table of bill_amount using city and connection_type.

```python
r=df2.pivot_table(index='city',columns='connection_type',values='bill_amount',aggfunc='sum',fill_value=0); print(r)
r.plot(kind='bar',figsize=(12,6),title='city vs connection_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q18. Entity Totals

Calculate total bill_amount by meter_id and display top 15.

```python
r=df2.groupby('meter_id')['bill_amount'].sum().nlargest(15); print(r)
r.plot(kind='bar',figsize=(12,5),title='Top 15 by bill_amount'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q19. Entity Frequency

Find the most frequent meter_id values.

```python
r=df2['meter_id'].value_counts().head(15); print(r)
r.plot(kind='bar',figsize=(11,5),title='Most Frequent meter_id'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q20. Inner Join

Merge the two DataFrames using meter_id and calculate bill_amount by city.

```python
j=df1.merge(df2,on='meter_id',how='inner'); r=j.groupby('city')['bill_amount'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Joined bill_amount by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q21. Join Row Comparison

Compare rows before and after the inner join.

```python
j=df1.merge(df2,on='meter_id',how='inner'); r=pd.Series(dict(DF1=len(df1),DF2=len(df2),Joined=len(j))); print(r)
r.plot(kind='bar',figsize=(7,5),title='Join Row Comparison'); plt.ylabel('Rows'); plt.show()
```

## Q22. Left Join Coverage

Measure matching versus unmatched meter_id records.

```python
j=df1.merge(df2,on='meter_id',how='left',indicator=True); r=j['_merge'].value_counts(); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Join Coverage'); plt.ylabel(''); plt.show()
```

## Q23. Average by Main Category

Calculate average bill_amount by city after joining.

```python
j=df1.merge(df2,on='meter_id',how='inner'); r=j.groupby('city')['bill_amount'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average bill_amount by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q24. Monthly Category Trend

Show monthly bill_amount trend for each connection_type.

```python
j=df2.copy(); j['month']=j['reading_date'].dt.to_period('M').astype(str); r=j.pivot_table(index='month',columns='connection_type',values='bill_amount',aggfunc='sum',fill_value=0); print(r.tail(12))
r.plot(figsize=(12,6),marker='o',title='Monthly bill_amount by connection_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q25. Top 5 Share

Find top 5 connection_type categories by total bill_amount.

```python
r=df2.groupby('connection_type')['bill_amount'].sum().nlargest(5); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Top 5 connection_type Share'); plt.ylabel(''); plt.show()
```

## Q26. Correlation

Calculate and visualize correlations among numeric columns.

```python
r=df2.select_dtypes('number').corr(); print(r)
plt.figure(figsize=(9,7)); plt.imshow(r,aspect='auto',cmap='viridis'); plt.colorbar(); plt.xticks(range(len(r)),r.columns,rotation=60); plt.yticks(range(len(r)),r.columns); plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

## Q27. KPI Summary

Create total, average, minimum and maximum KPIs for bill_amount.

```python
r=pd.Series(dict(Total=df2['bill_amount'].sum(),Average=df2['bill_amount'].mean(),Minimum=df2['bill_amount'].min(),Maximum=df2['bill_amount'].max())); print(r)
r.plot(kind='bar',figsize=(8,5),title='bill_amount KPI Summary'); plt.xticks(rotation=0); plt.show()
```

## Q28. Highest vs Lowest

Compare the highest and lowest connection_type by total bill_amount.

```python
r=df2.groupby('connection_type')['bill_amount'].sum().sort_values(ascending=False); r=r.iloc[[0,-1]]; print(r)
r.plot(kind='bar',figsize=(7,5),title='Highest vs Lowest connection_type'); plt.xticks(rotation=30); plt.show()
```

## Q29. Conditional Filter

Show records where bill_amount is above its overall average and visualize counts by connection_type.

```python
m=df2['bill_amount'].mean(); r=df2[df2['bill_amount']>m].groupby('connection_type').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Above-Average Records by connection_type'); plt.xticks(rotation=45); plt.show()
```

## Q30. Final Business Report

Create a final report by connection_type with count, total, average, minimum and maximum bill_amount.

```python
r=df2.groupby('connection_type').agg(records=('bill_amount','size'),total=('bill_amount','sum'),average=('bill_amount','mean'),minimum=('bill_amount','min'),maximum=('bill_amount','max')).sort_values('total',ascending=False); print(r)
r['total'].plot(kind='bar',figsize=(11,5),title='Final Business Report'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
