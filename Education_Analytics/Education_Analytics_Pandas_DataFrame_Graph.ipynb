# Pandas DataFrame + Graph Project — Education Analytics

**Input files:** `students.csv`, `exam_results.csv`

This project contains 30 business scenarios. **Every scenario includes a Pandas DataFrame analysis and a graph.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1=pd.read_csv('students.csv')
df2=pd.read_csv('exam_results.csv')
df2['exam_date']=pd.to_datetime(df2['exam_date'])

df1.head()
df2.head()
```

## Q1. Overview

Display shape, columns, data types and samples.

```python
print(df1.shape, df2.shape)
print(df1.dtypes)
print(df2.dtypes)
df1.head()
df2.head()
df2.select_dtypes('number').mean().plot(kind='bar',figsize=(9,4),title='Average Numeric Columns'); plt.tight_layout(); plt.show()
```

## Q2. Category Count

Count records by course.

```python
r=df2.groupby('course').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(9,5),title='Records by course'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q3. Category Total

Calculate total marks by subject.

```python
r=df2.groupby('subject')['marks'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Total marks by subject'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q4. Category Average

Calculate average marks by subject.

```python
r=df2.groupby('subject')['marks'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average marks by subject'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q5. Top 10

Find top 10 records by marks.

```python
r=df2.nlargest(10,'marks'); print(r)
r.set_index('student_id')['marks'].plot(kind='bar',figsize=(11,5),title='Top 10 marks'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q6. Bottom 10

Find bottom 10 records by marks.

```python
r=df2.nsmallest(10,'marks'); print(r)
r.set_index('student_id')['marks'].plot(kind='bar',figsize=(11,5),title='Bottom 10 marks'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q7. Missing Values

Calculate missing values in every column and visualize them.

```python
r=df2.isna().sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Missing Values'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q8. Duplicate Check

Count duplicate records and visualize unique versus duplicate rows.

```python
r=pd.Series({'Unique':len(df2)-df2.duplicated().sum(),'Duplicates':df2.duplicated().sum()}); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(6,6),title='Unique vs Duplicate'); plt.ylabel(''); plt.show()
```

## Q9. Distribution

Show descriptive statistics and distribution of marks.

```python
print(df2['marks'].describe())
df2['marks'].plot(kind='hist',bins=20,figsize=(9,5),title='Distribution of marks'); plt.xlabel('marks'); plt.tight_layout(); plt.show()
```

## Q10. Outliers

Identify marks outliers using IQR and visualize them.

```python
q1=df2['marks'].quantile(.25); q3=df2['marks'].quantile(.75); iqr=q3-q1
out=df2[(df2['marks']<q1-1.5*iqr)|(df2['marks']>q3+1.5*iqr)]; print('Outliers:',len(out))
df2.boxplot(column='marks',figsize=(7,5)); plt.title('marks Outliers'); plt.show()
```

## Q11. Monthly Trend

Calculate monthly total marks.

```python
df2['exam_date']=pd.to_datetime(df2['exam_date']); r=df2.groupby(df2['exam_date'].dt.to_period('M'))['marks'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(11,5),title='Monthly marks'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q12. Yearly Trend

Calculate yearly total marks.

```python
r=df2.groupby(df2['exam_date'].dt.year)['marks'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(9,5),title='Yearly marks'); plt.tight_layout(); plt.show()
```

## Q13. Percentage Share

Calculate percentage contribution of each subject to total marks.

```python
r=df2.groupby('subject')['marks'].sum(); r=r/r.sum()*100; print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='subject Share'); plt.ylabel(''); plt.show()
```

## Q14. Cumulative Trend

Calculate cumulative marks over time.

```python
r=df2.sort_values('exam_date').copy(); r['cumulative']=r['marks'].cumsum(); print(r[['exam_date','marks','cumulative']].head(20))
r.plot(x='exam_date',y='cumulative',figsize=(11,5),title='Cumulative marks'); plt.tight_layout(); plt.show()
```

## Q15. Category Ranking

Rank subject by total marks.

```python
r=df2.groupby('subject')['marks'].sum().sort_values(ascending=False).reset_index(); r['rank']=r['marks'].rank(method='dense',ascending=False); print(r)
r.plot(x='subject',y='marks',kind='bar',figsize=(10,5),title='Category Ranking',legend=False); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q16. Value Segments

Create Low, Medium and High segments for marks.

```python
m=df2['marks'].mean(); df2['segment']=np.select([df2['marks']<m*.7,df2['marks']<=m*1.3],['Low','Medium'],default='High'); r=df2['segment'].value_counts(); print(r)
r.plot(kind='bar',figsize=(8,5),title='Value Segments'); plt.tight_layout(); plt.show()
```

## Q17. Cross Tab

Create a pivot table of marks using course and subject.

```python
r=df2.pivot_table(index='course',columns='subject',values='marks',aggfunc='sum',fill_value=0); print(r)
r.plot(kind='bar',figsize=(12,6),title='course vs subject'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q18. Entity Totals

Calculate total marks by student_id and display top 15.

```python
r=df2.groupby('student_id')['marks'].sum().nlargest(15); print(r)
r.plot(kind='bar',figsize=(12,5),title='Top 15 by marks'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q19. Entity Frequency

Find the most frequent student_id values.

```python
r=df2['student_id'].value_counts().head(15); print(r)
r.plot(kind='bar',figsize=(11,5),title='Most Frequent student_id'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q20. Inner Join

Merge the two DataFrames using student_id and calculate marks by course.

```python
j=df1.merge(df2,on='student_id',how='inner'); r=j.groupby('course')['marks'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Joined marks by course'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q21. Join Row Comparison

Compare rows before and after the inner join.

```python
j=df1.merge(df2,on='student_id',how='inner'); r=pd.Series(dict(DF1=len(df1),DF2=len(df2),Joined=len(j))); print(r)
r.plot(kind='bar',figsize=(7,5),title='Join Row Comparison'); plt.ylabel('Rows'); plt.show()
```

## Q22. Left Join Coverage

Measure matching versus unmatched student_id records.

```python
j=df1.merge(df2,on='student_id',how='left',indicator=True); r=j['_merge'].value_counts(); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Join Coverage'); plt.ylabel(''); plt.show()
```

## Q23. Average by Main Category

Calculate average marks by course after joining.

```python
j=df1.merge(df2,on='student_id',how='inner'); r=j.groupby('course')['marks'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average marks by course'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q24. Monthly Category Trend

Show monthly marks trend for each subject.

```python
j=df2.copy(); j['month']=j['exam_date'].dt.to_period('M').astype(str); r=j.pivot_table(index='month',columns='subject',values='marks',aggfunc='sum',fill_value=0); print(r.tail(12))
r.plot(figsize=(12,6),marker='o',title='Monthly marks by subject'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q25. Top 5 Share

Find top 5 subject categories by total marks.

```python
r=df2.groupby('subject')['marks'].sum().nlargest(5); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Top 5 subject Share'); plt.ylabel(''); plt.show()
```

## Q26. Correlation

Calculate and visualize correlations among numeric columns.

```python
r=df2.select_dtypes('number').corr(); print(r)
plt.figure(figsize=(9,7)); plt.imshow(r,aspect='auto',cmap='viridis'); plt.colorbar(); plt.xticks(range(len(r)),r.columns,rotation=60); plt.yticks(range(len(r)),r.columns); plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

## Q27. KPI Summary

Create total, average, minimum and maximum KPIs for marks.

```python
r=pd.Series(dict(Total=df2['marks'].sum(),Average=df2['marks'].mean(),Minimum=df2['marks'].min(),Maximum=df2['marks'].max())); print(r)
r.plot(kind='bar',figsize=(8,5),title='marks KPI Summary'); plt.xticks(rotation=0); plt.show()
```

## Q28. Highest vs Lowest

Compare the highest and lowest subject by total marks.

```python
r=df2.groupby('subject')['marks'].sum().sort_values(ascending=False); r=r.iloc[[0,-1]]; print(r)
r.plot(kind='bar',figsize=(7,5),title='Highest vs Lowest subject'); plt.xticks(rotation=30); plt.show()
```

## Q29. Conditional Filter

Show records where marks is above its overall average and visualize counts by subject.

```python
m=df2['marks'].mean(); r=df2[df2['marks']>m].groupby('subject').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Above-Average Records by subject'); plt.xticks(rotation=45); plt.show()
```

## Q30. Final Business Report

Create a final report by subject with count, total, average, minimum and maximum marks.

```python
r=df2.groupby('subject').agg(records=('marks','size'),total=('marks','sum'),average=('marks','mean'),minimum=('marks','min'),maximum=('marks','max')).sort_values('total',ascending=False); print(r)
r['total'].plot(kind='bar',figsize=(11,5),title='Final Business Report'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
