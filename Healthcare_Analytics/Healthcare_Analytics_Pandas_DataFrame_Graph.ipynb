# Pandas DataFrame + Graph Project — Healthcare Analytics

**Input files:** `patients.csv`, `visits.csv`

This project contains 30 business scenarios. **Every scenario includes a Pandas DataFrame analysis and a graph.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1=pd.read_csv('patients.csv')
df2=pd.read_csv('visits.csv')
df2['visit_date']=pd.to_datetime(df2['visit_date'])

df1.head()
df2.head()
```

## Q1. Overview

Display shape, columns, data types and samples.

```python
print(df1.shape, df2.shape)
print(df1.dtypes)
print(df2.dtypes)
df1.head()
df2.head()
df2.select_dtypes('number').mean().plot(kind='bar',figsize=(9,4),title='Average Numeric Columns'); plt.tight_layout(); plt.show()
```

## Q2. Category Count

Count records by city.

```python
r=df2.groupby('city').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(9,5),title='Records by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q3. Category Total

Calculate total treatment_cost by department.

```python
r=df2.groupby('department')['treatment_cost'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Total treatment_cost by department'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q4. Category Average

Calculate average treatment_cost by department.

```python
r=df2.groupby('department')['treatment_cost'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average treatment_cost by department'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q5. Top 10

Find top 10 records by treatment_cost.

```python
r=df2.nlargest(10,'treatment_cost'); print(r)
r.set_index('patient_id')['treatment_cost'].plot(kind='bar',figsize=(11,5),title='Top 10 treatment_cost'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q6. Bottom 10

Find bottom 10 records by treatment_cost.

```python
r=df2.nsmallest(10,'treatment_cost'); print(r)
r.set_index('patient_id')['treatment_cost'].plot(kind='bar',figsize=(11,5),title='Bottom 10 treatment_cost'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q7. Missing Values

Calculate missing values in every column and visualize them.

```python
r=df2.isna().sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Missing Values'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q8. Duplicate Check

Count duplicate records and visualize unique versus duplicate rows.

```python
r=pd.Series({'Unique':len(df2)-df2.duplicated().sum(),'Duplicates':df2.duplicated().sum()}); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(6,6),title='Unique vs Duplicate'); plt.ylabel(''); plt.show()
```

## Q9. Distribution

Show descriptive statistics and distribution of treatment_cost.

```python
print(df2['treatment_cost'].describe())
df2['treatment_cost'].plot(kind='hist',bins=20,figsize=(9,5),title='Distribution of treatment_cost'); plt.xlabel('treatment_cost'); plt.tight_layout(); plt.show()
```

## Q10. Outliers

Identify treatment_cost outliers using IQR and visualize them.

```python
q1=df2['treatment_cost'].quantile(.25); q3=df2['treatment_cost'].quantile(.75); iqr=q3-q1
out=df2[(df2['treatment_cost']<q1-1.5*iqr)|(df2['treatment_cost']>q3+1.5*iqr)]; print('Outliers:',len(out))
df2.boxplot(column='treatment_cost',figsize=(7,5)); plt.title('treatment_cost Outliers'); plt.show()
```

## Q11. Monthly Trend

Calculate monthly total treatment_cost.

```python
df2['visit_date']=pd.to_datetime(df2['visit_date']); r=df2.groupby(df2['visit_date'].dt.to_period('M'))['treatment_cost'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(11,5),title='Monthly treatment_cost'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q12. Yearly Trend

Calculate yearly total treatment_cost.

```python
r=df2.groupby(df2['visit_date'].dt.year)['treatment_cost'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(9,5),title='Yearly treatment_cost'); plt.tight_layout(); plt.show()
```

## Q13. Percentage Share

Calculate percentage contribution of each department to total treatment_cost.

```python
r=df2.groupby('department')['treatment_cost'].sum(); r=r/r.sum()*100; print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='department Share'); plt.ylabel(''); plt.show()
```

## Q14. Cumulative Trend

Calculate cumulative treatment_cost over time.

```python
r=df2.sort_values('visit_date').copy(); r['cumulative']=r['treatment_cost'].cumsum(); print(r[['visit_date','treatment_cost','cumulative']].head(20))
r.plot(x='visit_date',y='cumulative',figsize=(11,5),title='Cumulative treatment_cost'); plt.tight_layout(); plt.show()
```

## Q15. Category Ranking

Rank department by total treatment_cost.

```python
r=df2.groupby('department')['treatment_cost'].sum().sort_values(ascending=False).reset_index(); r['rank']=r['treatment_cost'].rank(method='dense',ascending=False); print(r)
r.plot(x='department',y='treatment_cost',kind='bar',figsize=(10,5),title='Category Ranking',legend=False); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q16. Value Segments

Create Low, Medium and High segments for treatment_cost.

```python
m=df2['treatment_cost'].mean(); df2['segment']=np.select([df2['treatment_cost']<m*.7,df2['treatment_cost']<=m*1.3],['Low','Medium'],default='High'); r=df2['segment'].value_counts(); print(r)
r.plot(kind='bar',figsize=(8,5),title='Value Segments'); plt.tight_layout(); plt.show()
```

## Q17. Cross Tab

Create a pivot table of treatment_cost using city and department.

```python
r=df2.pivot_table(index='city',columns='department',values='treatment_cost',aggfunc='sum',fill_value=0); print(r)
r.plot(kind='bar',figsize=(12,6),title='city vs department'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q18. Entity Totals

Calculate total treatment_cost by patient_id and display top 15.

```python
r=df2.groupby('patient_id')['treatment_cost'].sum().nlargest(15); print(r)
r.plot(kind='bar',figsize=(12,5),title='Top 15 by treatment_cost'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q19. Entity Frequency

Find the most frequent patient_id values.

```python
r=df2['patient_id'].value_counts().head(15); print(r)
r.plot(kind='bar',figsize=(11,5),title='Most Frequent patient_id'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q20. Inner Join

Merge the two DataFrames using patient_id and calculate treatment_cost by city.

```python
j=df1.merge(df2,on='patient_id',how='inner'); r=j.groupby('city')['treatment_cost'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Joined treatment_cost by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q21. Join Row Comparison

Compare rows before and after the inner join.

```python
j=df1.merge(df2,on='patient_id',how='inner'); r=pd.Series(dict(DF1=len(df1),DF2=len(df2),Joined=len(j))); print(r)
r.plot(kind='bar',figsize=(7,5),title='Join Row Comparison'); plt.ylabel('Rows'); plt.show()
```

## Q22. Left Join Coverage

Measure matching versus unmatched patient_id records.

```python
j=df1.merge(df2,on='patient_id',how='left',indicator=True); r=j['_merge'].value_counts(); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Join Coverage'); plt.ylabel(''); plt.show()
```

## Q23. Average by Main Category

Calculate average treatment_cost by city after joining.

```python
j=df1.merge(df2,on='patient_id',how='inner'); r=j.groupby('city')['treatment_cost'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average treatment_cost by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q24. Monthly Category Trend

Show monthly treatment_cost trend for each department.

```python
j=df2.copy(); j['month']=j['visit_date'].dt.to_period('M').astype(str); r=j.pivot_table(index='month',columns='department',values='treatment_cost',aggfunc='sum',fill_value=0); print(r.tail(12))
r.plot(figsize=(12,6),marker='o',title='Monthly treatment_cost by department'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q25. Top 5 Share

Find top 5 department categories by total treatment_cost.

```python
r=df2.groupby('department')['treatment_cost'].sum().nlargest(5); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Top 5 department Share'); plt.ylabel(''); plt.show()
```

## Q26. Correlation

Calculate and visualize correlations among numeric columns.

```python
r=df2.select_dtypes('number').corr(); print(r)
plt.figure(figsize=(9,7)); plt.imshow(r,aspect='auto',cmap='viridis'); plt.colorbar(); plt.xticks(range(len(r)),r.columns,rotation=60); plt.yticks(range(len(r)),r.columns); plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

## Q27. KPI Summary

Create total, average, minimum and maximum KPIs for treatment_cost.

```python
r=pd.Series(dict(Total=df2['treatment_cost'].sum(),Average=df2['treatment_cost'].mean(),Minimum=df2['treatment_cost'].min(),Maximum=df2['treatment_cost'].max())); print(r)
r.plot(kind='bar',figsize=(8,5),title='treatment_cost KPI Summary'); plt.xticks(rotation=0); plt.show()
```

## Q28. Highest vs Lowest

Compare the highest and lowest department by total treatment_cost.

```python
r=df2.groupby('department')['treatment_cost'].sum().sort_values(ascending=False); r=r.iloc[[0,-1]]; print(r)
r.plot(kind='bar',figsize=(7,5),title='Highest vs Lowest department'); plt.xticks(rotation=30); plt.show()
```

## Q29. Conditional Filter

Show records where treatment_cost is above its overall average and visualize counts by department.

```python
m=df2['treatment_cost'].mean(); r=df2[df2['treatment_cost']>m].groupby('department').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Above-Average Records by department'); plt.xticks(rotation=45); plt.show()
```

## Q30. Final Business Report

Create a final report by department with count, total, average, minimum and maximum treatment_cost.

```python
r=df2.groupby('department').agg(records=('treatment_cost','size'),total=('treatment_cost','sum'),average=('treatment_cost','mean'),minimum=('treatment_cost','min'),maximum=('treatment_cost','max')).sort_values('total',ascending=False); print(r)
r['total'].plot(kind='bar',figsize=(11,5),title='Final Business Report'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
