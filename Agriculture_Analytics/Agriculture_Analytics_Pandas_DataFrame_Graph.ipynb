# Pandas DataFrame + Graph Project — Agriculture Analytics

**Input files:** `farms.csv`, `harvest_records.csv`

This project contains 30 business scenarios. **Every scenario includes a Pandas DataFrame analysis and a graph.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1=pd.read_csv('farms.csv')
df2=pd.read_csv('harvest_records.csv')
df2['harvest_date']=pd.to_datetime(df2['harvest_date'])

df1.head()
df2.head()
```

## Q1. Overview

Display shape, columns, data types and samples.

```python
print(df1.shape, df2.shape)
print(df1.dtypes)
print(df2.dtypes)
df1.head()
df2.head()
df2.select_dtypes('number').mean().plot(kind='bar',figsize=(9,4),title='Average Numeric Columns'); plt.tight_layout(); plt.show()
```

## Q2. Category Count

Count records by district.

```python
r=df2.groupby('district').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(9,5),title='Records by district'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q3. Category Total

Calculate total revenue by crop_type.

```python
r=df2.groupby('crop_type')['revenue'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Total revenue by crop_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q4. Category Average

Calculate average revenue by crop_type.

```python
r=df2.groupby('crop_type')['revenue'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average revenue by crop_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q5. Top 10

Find top 10 records by revenue.

```python
r=df2.nlargest(10,'revenue'); print(r)
r.set_index('farm_id')['revenue'].plot(kind='bar',figsize=(11,5),title='Top 10 revenue'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q6. Bottom 10

Find bottom 10 records by revenue.

```python
r=df2.nsmallest(10,'revenue'); print(r)
r.set_index('farm_id')['revenue'].plot(kind='bar',figsize=(11,5),title='Bottom 10 revenue'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q7. Missing Values

Calculate missing values in every column and visualize them.

```python
r=df2.isna().sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Missing Values'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q8. Duplicate Check

Count duplicate records and visualize unique versus duplicate rows.

```python
r=pd.Series({'Unique':len(df2)-df2.duplicated().sum(),'Duplicates':df2.duplicated().sum()}); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(6,6),title='Unique vs Duplicate'); plt.ylabel(''); plt.show()
```

## Q9. Distribution

Show descriptive statistics and distribution of revenue.

```python
print(df2['revenue'].describe())
df2['revenue'].plot(kind='hist',bins=20,figsize=(9,5),title='Distribution of revenue'); plt.xlabel('revenue'); plt.tight_layout(); plt.show()
```

## Q10. Outliers

Identify revenue outliers using IQR and visualize them.

```python
q1=df2['revenue'].quantile(.25); q3=df2['revenue'].quantile(.75); iqr=q3-q1
out=df2[(df2['revenue']<q1-1.5*iqr)|(df2['revenue']>q3+1.5*iqr)]; print('Outliers:',len(out))
df2.boxplot(column='revenue',figsize=(7,5)); plt.title('revenue Outliers'); plt.show()
```

## Q11. Monthly Trend

Calculate monthly total revenue.

```python
df2['harvest_date']=pd.to_datetime(df2['harvest_date']); r=df2.groupby(df2['harvest_date'].dt.to_period('M'))['revenue'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(11,5),title='Monthly revenue'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q12. Yearly Trend

Calculate yearly total revenue.

```python
r=df2.groupby(df2['harvest_date'].dt.year)['revenue'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(9,5),title='Yearly revenue'); plt.tight_layout(); plt.show()
```

## Q13. Percentage Share

Calculate percentage contribution of each crop_type to total revenue.

```python
r=df2.groupby('crop_type')['revenue'].sum(); r=r/r.sum()*100; print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='crop_type Share'); plt.ylabel(''); plt.show()
```

## Q14. Cumulative Trend

Calculate cumulative revenue over time.

```python
r=df2.sort_values('harvest_date').copy(); r['cumulative']=r['revenue'].cumsum(); print(r[['harvest_date','revenue','cumulative']].head(20))
r.plot(x='harvest_date',y='cumulative',figsize=(11,5),title='Cumulative revenue'); plt.tight_layout(); plt.show()
```

## Q15. Category Ranking

Rank crop_type by total revenue.

```python
r=df2.groupby('crop_type')['revenue'].sum().sort_values(ascending=False).reset_index(); r['rank']=r['revenue'].rank(method='dense',ascending=False); print(r)
r.plot(x='crop_type',y='revenue',kind='bar',figsize=(10,5),title='Category Ranking',legend=False); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q16. Value Segments

Create Low, Medium and High segments for revenue.

```python
m=df2['revenue'].mean(); df2['segment']=np.select([df2['revenue']<m*.7,df2['revenue']<=m*1.3],['Low','Medium'],default='High'); r=df2['segment'].value_counts(); print(r)
r.plot(kind='bar',figsize=(8,5),title='Value Segments'); plt.tight_layout(); plt.show()
```

## Q17. Cross Tab

Create a pivot table of revenue using district and crop_type.

```python
r=df2.pivot_table(index='district',columns='crop_type',values='revenue',aggfunc='sum',fill_value=0); print(r)
r.plot(kind='bar',figsize=(12,6),title='district vs crop_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q18. Entity Totals

Calculate total revenue by farm_id and display top 15.

```python
r=df2.groupby('farm_id')['revenue'].sum().nlargest(15); print(r)
r.plot(kind='bar',figsize=(12,5),title='Top 15 by revenue'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q19. Entity Frequency

Find the most frequent farm_id values.

```python
r=df2['farm_id'].value_counts().head(15); print(r)
r.plot(kind='bar',figsize=(11,5),title='Most Frequent farm_id'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q20. Inner Join

Merge the two DataFrames using farm_id and calculate revenue by district.

```python
j=df1.merge(df2,on='farm_id',how='inner'); r=j.groupby('district')['revenue'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Joined revenue by district'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q21. Join Row Comparison

Compare rows before and after the inner join.

```python
j=df1.merge(df2,on='farm_id',how='inner'); r=pd.Series(dict(DF1=len(df1),DF2=len(df2),Joined=len(j))); print(r)
r.plot(kind='bar',figsize=(7,5),title='Join Row Comparison'); plt.ylabel('Rows'); plt.show()
```

## Q22. Left Join Coverage

Measure matching versus unmatched farm_id records.

```python
j=df1.merge(df2,on='farm_id',how='left',indicator=True); r=j['_merge'].value_counts(); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Join Coverage'); plt.ylabel(''); plt.show()
```

## Q23. Average by Main Category

Calculate average revenue by district after joining.

```python
j=df1.merge(df2,on='farm_id',how='inner'); r=j.groupby('district')['revenue'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average revenue by district'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q24. Monthly Category Trend

Show monthly revenue trend for each crop_type.

```python
j=df2.copy(); j['month']=j['harvest_date'].dt.to_period('M').astype(str); r=j.pivot_table(index='month',columns='crop_type',values='revenue',aggfunc='sum',fill_value=0); print(r.tail(12))
r.plot(figsize=(12,6),marker='o',title='Monthly revenue by crop_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q25. Top 5 Share

Find top 5 crop_type categories by total revenue.

```python
r=df2.groupby('crop_type')['revenue'].sum().nlargest(5); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Top 5 crop_type Share'); plt.ylabel(''); plt.show()
```

## Q26. Correlation

Calculate and visualize correlations among numeric columns.

```python
r=df2.select_dtypes('number').corr(); print(r)
plt.figure(figsize=(9,7)); plt.imshow(r,aspect='auto',cmap='viridis'); plt.colorbar(); plt.xticks(range(len(r)),r.columns,rotation=60); plt.yticks(range(len(r)),r.columns); plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

## Q27. KPI Summary

Create total, average, minimum and maximum KPIs for revenue.

```python
r=pd.Series(dict(Total=df2['revenue'].sum(),Average=df2['revenue'].mean(),Minimum=df2['revenue'].min(),Maximum=df2['revenue'].max())); print(r)
r.plot(kind='bar',figsize=(8,5),title='revenue KPI Summary'); plt.xticks(rotation=0); plt.show()
```

## Q28. Highest vs Lowest

Compare the highest and lowest crop_type by total revenue.

```python
r=df2.groupby('crop_type')['revenue'].sum().sort_values(ascending=False); r=r.iloc[[0,-1]]; print(r)
r.plot(kind='bar',figsize=(7,5),title='Highest vs Lowest crop_type'); plt.xticks(rotation=30); plt.show()
```

## Q29. Conditional Filter

Show records where revenue is above its overall average and visualize counts by crop_type.

```python
m=df2['revenue'].mean(); r=df2[df2['revenue']>m].groupby('crop_type').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Above-Average Records by crop_type'); plt.xticks(rotation=45); plt.show()
```

## Q30. Final Business Report

Create a final report by crop_type with count, total, average, minimum and maximum revenue.

```python
r=df2.groupby('crop_type').agg(records=('revenue','size'),total=('revenue','sum'),average=('revenue','mean'),minimum=('revenue','min'),maximum=('revenue','max')).sort_values('total',ascending=False); print(r)
r['total'].plot(kind='bar',figsize=(11,5),title='Final Business Report'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
