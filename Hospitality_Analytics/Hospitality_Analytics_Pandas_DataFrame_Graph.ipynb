# Pandas DataFrame + Graph Project — Hospitality Analytics

**Input files:** `hotels.csv`, `hotel_bookings.csv`

This project contains 30 business scenarios. **Every scenario includes a Pandas DataFrame analysis and a graph.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1=pd.read_csv('hotels.csv')
df2=pd.read_csv('hotel_bookings.csv')
df2['booking_date']=pd.to_datetime(df2['booking_date'])

df1.head()
df2.head()
```

## Q1. Overview

Display shape, columns, data types and samples.

```python
print(df1.shape, df2.shape)
print(df1.dtypes)
print(df2.dtypes)
df1.head()
df2.head()
df2.select_dtypes('number').mean().plot(kind='bar',figsize=(9,4),title='Average Numeric Columns'); plt.tight_layout(); plt.show()
```

## Q2. Category Count

Count records by city.

```python
r=df2.groupby('city').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(9,5),title='Records by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q3. Category Total

Calculate total room_charge by room_type.

```python
r=df2.groupby('room_type')['room_charge'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Total room_charge by room_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q4. Category Average

Calculate average room_charge by room_type.

```python
r=df2.groupby('room_type')['room_charge'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average room_charge by room_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q5. Top 10

Find top 10 records by room_charge.

```python
r=df2.nlargest(10,'room_charge'); print(r)
r.set_index('hotel_id')['room_charge'].plot(kind='bar',figsize=(11,5),title='Top 10 room_charge'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q6. Bottom 10

Find bottom 10 records by room_charge.

```python
r=df2.nsmallest(10,'room_charge'); print(r)
r.set_index('hotel_id')['room_charge'].plot(kind='bar',figsize=(11,5),title='Bottom 10 room_charge'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q7. Missing Values

Calculate missing values in every column and visualize them.

```python
r=df2.isna().sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Missing Values'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q8. Duplicate Check

Count duplicate records and visualize unique versus duplicate rows.

```python
r=pd.Series({'Unique':len(df2)-df2.duplicated().sum(),'Duplicates':df2.duplicated().sum()}); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(6,6),title='Unique vs Duplicate'); plt.ylabel(''); plt.show()
```

## Q9. Distribution

Show descriptive statistics and distribution of room_charge.

```python
print(df2['room_charge'].describe())
df2['room_charge'].plot(kind='hist',bins=20,figsize=(9,5),title='Distribution of room_charge'); plt.xlabel('room_charge'); plt.tight_layout(); plt.show()
```

## Q10. Outliers

Identify room_charge outliers using IQR and visualize them.

```python
q1=df2['room_charge'].quantile(.25); q3=df2['room_charge'].quantile(.75); iqr=q3-q1
out=df2[(df2['room_charge']<q1-1.5*iqr)|(df2['room_charge']>q3+1.5*iqr)]; print('Outliers:',len(out))
df2.boxplot(column='room_charge',figsize=(7,5)); plt.title('room_charge Outliers'); plt.show()
```

## Q11. Monthly Trend

Calculate monthly total room_charge.

```python
df2['booking_date']=pd.to_datetime(df2['booking_date']); r=df2.groupby(df2['booking_date'].dt.to_period('M'))['room_charge'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(11,5),title='Monthly room_charge'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q12. Yearly Trend

Calculate yearly total room_charge.

```python
r=df2.groupby(df2['booking_date'].dt.year)['room_charge'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(9,5),title='Yearly room_charge'); plt.tight_layout(); plt.show()
```

## Q13. Percentage Share

Calculate percentage contribution of each room_type to total room_charge.

```python
r=df2.groupby('room_type')['room_charge'].sum(); r=r/r.sum()*100; print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='room_type Share'); plt.ylabel(''); plt.show()
```

## Q14. Cumulative Trend

Calculate cumulative room_charge over time.

```python
r=df2.sort_values('booking_date').copy(); r['cumulative']=r['room_charge'].cumsum(); print(r[['booking_date','room_charge','cumulative']].head(20))
r.plot(x='booking_date',y='cumulative',figsize=(11,5),title='Cumulative room_charge'); plt.tight_layout(); plt.show()
```

## Q15. Category Ranking

Rank room_type by total room_charge.

```python
r=df2.groupby('room_type')['room_charge'].sum().sort_values(ascending=False).reset_index(); r['rank']=r['room_charge'].rank(method='dense',ascending=False); print(r)
r.plot(x='room_type',y='room_charge',kind='bar',figsize=(10,5),title='Category Ranking',legend=False); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q16. Value Segments

Create Low, Medium and High segments for room_charge.

```python
m=df2['room_charge'].mean(); df2['segment']=np.select([df2['room_charge']<m*.7,df2['room_charge']<=m*1.3],['Low','Medium'],default='High'); r=df2['segment'].value_counts(); print(r)
r.plot(kind='bar',figsize=(8,5),title='Value Segments'); plt.tight_layout(); plt.show()
```

## Q17. Cross Tab

Create a pivot table of room_charge using city and room_type.

```python
r=df2.pivot_table(index='city',columns='room_type',values='room_charge',aggfunc='sum',fill_value=0); print(r)
r.plot(kind='bar',figsize=(12,6),title='city vs room_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q18. Entity Totals

Calculate total room_charge by hotel_id and display top 15.

```python
r=df2.groupby('hotel_id')['room_charge'].sum().nlargest(15); print(r)
r.plot(kind='bar',figsize=(12,5),title='Top 15 by room_charge'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q19. Entity Frequency

Find the most frequent hotel_id values.

```python
r=df2['hotel_id'].value_counts().head(15); print(r)
r.plot(kind='bar',figsize=(11,5),title='Most Frequent hotel_id'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q20. Inner Join

Merge the two DataFrames using hotel_id and calculate room_charge by city.

```python
j=df1.merge(df2,on='hotel_id',how='inner'); r=j.groupby('city')['room_charge'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Joined room_charge by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q21. Join Row Comparison

Compare rows before and after the inner join.

```python
j=df1.merge(df2,on='hotel_id',how='inner'); r=pd.Series(dict(DF1=len(df1),DF2=len(df2),Joined=len(j))); print(r)
r.plot(kind='bar',figsize=(7,5),title='Join Row Comparison'); plt.ylabel('Rows'); plt.show()
```

## Q22. Left Join Coverage

Measure matching versus unmatched hotel_id records.

```python
j=df1.merge(df2,on='hotel_id',how='left',indicator=True); r=j['_merge'].value_counts(); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Join Coverage'); plt.ylabel(''); plt.show()
```

## Q23. Average by Main Category

Calculate average room_charge by city after joining.

```python
j=df1.merge(df2,on='hotel_id',how='inner'); r=j.groupby('city')['room_charge'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average room_charge by city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q24. Monthly Category Trend

Show monthly room_charge trend for each room_type.

```python
j=df2.copy(); j['month']=j['booking_date'].dt.to_period('M').astype(str); r=j.pivot_table(index='month',columns='room_type',values='room_charge',aggfunc='sum',fill_value=0); print(r.tail(12))
r.plot(figsize=(12,6),marker='o',title='Monthly room_charge by room_type'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q25. Top 5 Share

Find top 5 room_type categories by total room_charge.

```python
r=df2.groupby('room_type')['room_charge'].sum().nlargest(5); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Top 5 room_type Share'); plt.ylabel(''); plt.show()
```

## Q26. Correlation

Calculate and visualize correlations among numeric columns.

```python
r=df2.select_dtypes('number').corr(); print(r)
plt.figure(figsize=(9,7)); plt.imshow(r,aspect='auto',cmap='viridis'); plt.colorbar(); plt.xticks(range(len(r)),r.columns,rotation=60); plt.yticks(range(len(r)),r.columns); plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

## Q27. KPI Summary

Create total, average, minimum and maximum KPIs for room_charge.

```python
r=pd.Series(dict(Total=df2['room_charge'].sum(),Average=df2['room_charge'].mean(),Minimum=df2['room_charge'].min(),Maximum=df2['room_charge'].max())); print(r)
r.plot(kind='bar',figsize=(8,5),title='room_charge KPI Summary'); plt.xticks(rotation=0); plt.show()
```

## Q28. Highest vs Lowest

Compare the highest and lowest room_type by total room_charge.

```python
r=df2.groupby('room_type')['room_charge'].sum().sort_values(ascending=False); r=r.iloc[[0,-1]]; print(r)
r.plot(kind='bar',figsize=(7,5),title='Highest vs Lowest room_type'); plt.xticks(rotation=30); plt.show()
```

## Q29. Conditional Filter

Show records where room_charge is above its overall average and visualize counts by room_type.

```python
m=df2['room_charge'].mean(); r=df2[df2['room_charge']>m].groupby('room_type').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Above-Average Records by room_type'); plt.xticks(rotation=45); plt.show()
```

## Q30. Final Business Report

Create a final report by room_type with count, total, average, minimum and maximum room_charge.

```python
r=df2.groupby('room_type').agg(records=('room_charge','size'),total=('room_charge','sum'),average=('room_charge','mean'),minimum=('room_charge','min'),maximum=('room_charge','max')).sort_values('total',ascending=False); print(r)
r['total'].plot(kind='bar',figsize=(11,5),title='Final Business Report'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
