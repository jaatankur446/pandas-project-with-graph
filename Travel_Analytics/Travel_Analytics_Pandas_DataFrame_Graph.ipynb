# Pandas DataFrame + Graph Project — Travel Analytics

**Input files:** `travelers.csv`, `bookings.csv`

This project contains 30 business scenarios. **Every scenario includes a Pandas DataFrame analysis and a graph.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df1=pd.read_csv('travelers.csv')
df2=pd.read_csv('bookings.csv')
df2['booking_date']=pd.to_datetime(df2['booking_date'])

df1.head()
df2.head()
```

## Q1. Overview

Display shape, columns, data types and samples.

```python
print(df1.shape, df2.shape)
print(df1.dtypes)
print(df2.dtypes)
df1.head()
df2.head()
df2.select_dtypes('number').mean().plot(kind='bar',figsize=(9,4),title='Average Numeric Columns'); plt.tight_layout(); plt.show()
```

## Q2. Category Count

Count records by home_city.

```python
r=df2.groupby('home_city').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(9,5),title='Records by home_city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q3. Category Total

Calculate total fare by destination.

```python
r=df2.groupby('destination')['fare'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Total fare by destination'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q4. Category Average

Calculate average fare by destination.

```python
r=df2.groupby('destination')['fare'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average fare by destination'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q5. Top 10

Find top 10 records by fare.

```python
r=df2.nlargest(10,'fare'); print(r)
r.set_index('traveler_id')['fare'].plot(kind='bar',figsize=(11,5),title='Top 10 fare'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q6. Bottom 10

Find bottom 10 records by fare.

```python
r=df2.nsmallest(10,'fare'); print(r)
r.set_index('traveler_id')['fare'].plot(kind='bar',figsize=(11,5),title='Bottom 10 fare'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q7. Missing Values

Calculate missing values in every column and visualize them.

```python
r=df2.isna().sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Missing Values'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q8. Duplicate Check

Count duplicate records and visualize unique versus duplicate rows.

```python
r=pd.Series({'Unique':len(df2)-df2.duplicated().sum(),'Duplicates':df2.duplicated().sum()}); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(6,6),title='Unique vs Duplicate'); plt.ylabel(''); plt.show()
```

## Q9. Distribution

Show descriptive statistics and distribution of fare.

```python
print(df2['fare'].describe())
df2['fare'].plot(kind='hist',bins=20,figsize=(9,5),title='Distribution of fare'); plt.xlabel('fare'); plt.tight_layout(); plt.show()
```

## Q10. Outliers

Identify fare outliers using IQR and visualize them.

```python
q1=df2['fare'].quantile(.25); q3=df2['fare'].quantile(.75); iqr=q3-q1
out=df2[(df2['fare']<q1-1.5*iqr)|(df2['fare']>q3+1.5*iqr)]; print('Outliers:',len(out))
df2.boxplot(column='fare',figsize=(7,5)); plt.title('fare Outliers'); plt.show()
```

## Q11. Monthly Trend

Calculate monthly total fare.

```python
df2['booking_date']=pd.to_datetime(df2['booking_date']); r=df2.groupby(df2['booking_date'].dt.to_period('M'))['fare'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(11,5),title='Monthly fare'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q12. Yearly Trend

Calculate yearly total fare.

```python
r=df2.groupby(df2['booking_date'].dt.year)['fare'].sum(); print(r)
r.plot(kind='line',marker='o',figsize=(9,5),title='Yearly fare'); plt.tight_layout(); plt.show()
```

## Q13. Percentage Share

Calculate percentage contribution of each destination to total fare.

```python
r=df2.groupby('destination')['fare'].sum(); r=r/r.sum()*100; print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='destination Share'); plt.ylabel(''); plt.show()
```

## Q14. Cumulative Trend

Calculate cumulative fare over time.

```python
r=df2.sort_values('booking_date').copy(); r['cumulative']=r['fare'].cumsum(); print(r[['booking_date','fare','cumulative']].head(20))
r.plot(x='booking_date',y='cumulative',figsize=(11,5),title='Cumulative fare'); plt.tight_layout(); plt.show()
```

## Q15. Category Ranking

Rank destination by total fare.

```python
r=df2.groupby('destination')['fare'].sum().sort_values(ascending=False).reset_index(); r['rank']=r['fare'].rank(method='dense',ascending=False); print(r)
r.plot(x='destination',y='fare',kind='bar',figsize=(10,5),title='Category Ranking',legend=False); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q16. Value Segments

Create Low, Medium and High segments for fare.

```python
m=df2['fare'].mean(); df2['segment']=np.select([df2['fare']<m*.7,df2['fare']<=m*1.3],['Low','Medium'],default='High'); r=df2['segment'].value_counts(); print(r)
r.plot(kind='bar',figsize=(8,5),title='Value Segments'); plt.tight_layout(); plt.show()
```

## Q17. Cross Tab

Create a pivot table of fare using home_city and destination.

```python
r=df2.pivot_table(index='home_city',columns='destination',values='fare',aggfunc='sum',fill_value=0); print(r)
r.plot(kind='bar',figsize=(12,6),title='home_city vs destination'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q18. Entity Totals

Calculate total fare by traveler_id and display top 15.

```python
r=df2.groupby('traveler_id')['fare'].sum().nlargest(15); print(r)
r.plot(kind='bar',figsize=(12,5),title='Top 15 by fare'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q19. Entity Frequency

Find the most frequent traveler_id values.

```python
r=df2['traveler_id'].value_counts().head(15); print(r)
r.plot(kind='bar',figsize=(11,5),title='Most Frequent traveler_id'); plt.xticks(rotation=60); plt.tight_layout(); plt.show()
```

## Q20. Inner Join

Merge the two DataFrames using traveler_id and calculate fare by home_city.

```python
j=df1.merge(df2,on='traveler_id',how='inner'); r=j.groupby('home_city')['fare'].sum().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Joined fare by home_city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q21. Join Row Comparison

Compare rows before and after the inner join.

```python
j=df1.merge(df2,on='traveler_id',how='inner'); r=pd.Series(dict(DF1=len(df1),DF2=len(df2),Joined=len(j))); print(r)
r.plot(kind='bar',figsize=(7,5),title='Join Row Comparison'); plt.ylabel('Rows'); plt.show()
```

## Q22. Left Join Coverage

Measure matching versus unmatched traveler_id records.

```python
j=df1.merge(df2,on='traveler_id',how='left',indicator=True); r=j['_merge'].value_counts(); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Join Coverage'); plt.ylabel(''); plt.show()
```

## Q23. Average by Main Category

Calculate average fare by home_city after joining.

```python
j=df1.merge(df2,on='traveler_id',how='inner'); r=j.groupby('home_city')['fare'].mean().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Average fare by home_city'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q24. Monthly Category Trend

Show monthly fare trend for each destination.

```python
j=df2.copy(); j['month']=j['booking_date'].dt.to_period('M').astype(str); r=j.pivot_table(index='month',columns='destination',values='fare',aggfunc='sum',fill_value=0); print(r.tail(12))
r.plot(figsize=(12,6),marker='o',title='Monthly fare by destination'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

## Q25. Top 5 Share

Find top 5 destination categories by total fare.

```python
r=df2.groupby('destination')['fare'].sum().nlargest(5); print(r)
r.plot(kind='pie',autopct='%1.1f%%',figsize=(7,7),title='Top 5 destination Share'); plt.ylabel(''); plt.show()
```

## Q26. Correlation

Calculate and visualize correlations among numeric columns.

```python
r=df2.select_dtypes('number').corr(); print(r)
plt.figure(figsize=(9,7)); plt.imshow(r,aspect='auto',cmap='viridis'); plt.colorbar(); plt.xticks(range(len(r)),r.columns,rotation=60); plt.yticks(range(len(r)),r.columns); plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

## Q27. KPI Summary

Create total, average, minimum and maximum KPIs for fare.

```python
r=pd.Series(dict(Total=df2['fare'].sum(),Average=df2['fare'].mean(),Minimum=df2['fare'].min(),Maximum=df2['fare'].max())); print(r)
r.plot(kind='bar',figsize=(8,5),title='fare KPI Summary'); plt.xticks(rotation=0); plt.show()
```

## Q28. Highest vs Lowest

Compare the highest and lowest destination by total fare.

```python
r=df2.groupby('destination')['fare'].sum().sort_values(ascending=False); r=r.iloc[[0,-1]]; print(r)
r.plot(kind='bar',figsize=(7,5),title='Highest vs Lowest destination'); plt.xticks(rotation=30); plt.show()
```

## Q29. Conditional Filter

Show records where fare is above its overall average and visualize counts by destination.

```python
m=df2['fare'].mean(); r=df2[df2['fare']>m].groupby('destination').size().sort_values(ascending=False); print(r)
r.plot(kind='bar',figsize=(10,5),title='Above-Average Records by destination'); plt.xticks(rotation=45); plt.show()
```

## Q30. Final Business Report

Create a final report by destination with count, total, average, minimum and maximum fare.

```python
r=df2.groupby('destination').agg(records=('fare','size'),total=('fare','sum'),average=('fare','mean'),minimum=('fare','min'),maximum=('fare','max')).sort_values('total',ascending=False); print(r)
r['total'].plot(kind='bar',figsize=(11,5),title='Final Business Report'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```
